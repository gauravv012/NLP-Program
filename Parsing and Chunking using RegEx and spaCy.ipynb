## Resources: Parsing and Chunking using RegEx and spaCy

This notebook will demonstrate basic text parsing and chunking techniques using Python's built-in `re` module for regular expressions and the `spaCy` library.

```python
# Install spaCy (if not already installed)
# !pip install spacy
# Download a spaCy language model (e.g., 'en_core_web_sm')
# !python -m spacy download en_core_web_sm

import re
import spacy

print("Libraries imported successfully.")
```

### 1. Parsing and Chunking with Regular Expressions (RegEx)

Regular expressions are powerful for pattern matching in text. Here, we'll extract specific patterns (like email addresses or sentences) from a sample text.

```python
sample_text = """Hello there! My email is example@domain.com. You can also reach out to info@anothersite.org. This is the first sentence. This is the second sentence, and it has a comma. Finally, the third one.
"""

# Example 1: Extracting email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, sample_text)
print("Extracted Emails:", emails)

# Example 2: Sentence chunking
sentence_pattern = r'[^.!?]+[.!?]'
sentences = re.findall(sentence_pattern, sample_text)
print("\nExtracted Sentences:")
for i, sent in enumerate(sentences):
    print(f"{i+1}. {sent.strip()}")
```

### 2. Parsing and Chunking with spaCy

spaCy is an industrial-strength natural language processing (NLP) library. It provides pre-trained models for various tasks, including tokenization, sentence segmentation, part-of-speech tagging, named entity recognition, and more. It's excellent for more sophisticated text analysis.

First, we need to load a spaCy model.

```python
# Load the English language model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading en_core_web_sm model...")
    !python -m spacy download en_core_web_sm
    nlp = spacy.load("en_core_web_sm")

print("spaCy model loaded successfully.")
```

Now, let's process our sample text using spaCy to perform various parsing and chunking operations.

```python
doc = nlp(sample_text)

print("\n--- spaCy Tokenization ---")
for token in doc:
    print(f"Token: '{token.text}', Lemma: '{token.lemma_}', POS: '{token.pos_}', Dep: '{token.dep_}'")

print("\n--- spaCy Sentence Segmentation ---")
for i, sent in enumerate(doc.sents):
    print(f"{i+1}. {sent.text}")

print("\n--- spaCy Noun Chunks (Chunking) ---")
for chunk in doc.noun_chunks:
    print(f"Noun Chunk: '{chunk.text}', Root: '{chunk.root.text}', Root POS: '{chunk.root.pos_}'")

print("\n--- spaCy Named Entity Recognition (NER) ---")
for ent in doc.ents:
    print(f"Entity: '{ent.text}', Label: '{ent.label_}'")
```
