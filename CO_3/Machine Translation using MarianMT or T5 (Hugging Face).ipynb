```python
from transformers import MarianMTModel, MarianTokenizer

# English -> Hindi (use "Helsinki-NLP/opus-mt-en-fr" for French, "opus-mt-en-de" for German, etc.)
MODEL_NAME = "Helsinki-NLP/opus-mt-en-hi"

tokenizer = MarianTokenizer.from_pretrained(MODEL_NAME)
model = MarianMTModel.from_pretrained(MODEL_NAME)

sentences = [
    "Machine translation is a key application of natural language processing.",
    "How are you today?",
    "I am learning transformers in my NLP lab."
]

inputs = tokenizer(sentences, return_tensors="pt", padding=True, truncation=True)
translated = model.generate(**inputs, max_length=128, num_beams=4)

for src, tgt in zip(sentences, translated):
    print("English :", src)
    print("Hindi   :", tokenizer.decode(tgt, skip_special_tokens=True))
    print()
```
