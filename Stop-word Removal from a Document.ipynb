## Stop Word Removal

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download the stopwords corpus if you haven't already
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

print("NLTK stopwords and punkt tokenizer are ready.")
```

Let's define a sample document to demonstrate stop word removal.

```python
sample_document = "This is an example sentence, showing off the stop word removal process. It is quite interesting and useful for text preprocessing."
print("Original Document:")
print(sample_document)
```

Now, we will tokenize the document and remove the stop words.

```python
# Get the English stop words
stop_words = set(stopwords.words('english'))

# Download punkt_tab if not already present, as it's needed by word_tokenize
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

# Tokenize the document
word_tokens = word_tokenize(sample_document)

# Filter out stop words
filtered_sentence = [w for w in word_tokens if not w.lower() in stop_words]

# Join the filtered words back into a sentence (optional, for display)
filtered_document = " ".join(filtered_sentence)

print("\nDocument after Stop Word Removal:")
print(filtered_document)
```
