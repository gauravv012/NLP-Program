```python
!python -m spacy download en_core_web_sm
```

Now, we can load the model and use it to perform Named Entity Recognition on a sample text.

```python
import spacy

# Load the English NLP model
nlp = spacy.load('en_core_web_sm')

# Sample text for NER
text = "Apple Inc. was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne on April 1, 1976, in Cupertino, California. It became a public company in 1980."

# Process the text with the NLP model
doc = nlp(text)

# Print identified entities and their labels
print("Named Entities:")
for ent in doc.ents:
    print(f"- {ent.text} ({ent.label_})")
```
