## Part-of-Speech Tagging

```python
# Install nltk if not already installed (uncomment and run if needed)
# !pip install nltk

import nltk

# Download necessary NLTK data (punkt for tokenization, averaged_perceptron_tagger for POS tagging)
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
```

```python
# Define the sentence to be tagged
sentence = "The quick brown fox jumps over the lazy dog."
print(f"Original Sentence: {sentence}")
```

```python
# Tokenize the sentence into words
nltk.download('punkt_tab') # Download missing resource
words = nltk.word_tokenize(sentence)
print(f"Tokenized Words: {words}")
```

```python
nltk.download('averaged_perceptron_tagger_eng')
# Perform Part-of-Speech Tagging
pos_tags = nltk.pos_tag(words)

# Display the results
print(f"Part-of-Speech Tags: {pos_tags}")
```

The output shows each word paired with its corresponding part-of-speech tag:

*   `DT`: Determiner
*   `JJ`: Adjective
*   `NN`: Noun, singular or mass
*   `VBZ`: Verb, third person singular present
*   `IN`: Preposition or subordinating conjunction
