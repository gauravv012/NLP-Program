## Sample Text for Tokenization

```python
sample_text = (
    "NLTK is a powerful library for working with human language data. "
    "SpaCy, on the other hand, is designed for production use and efficiency. "
    "Both are excellent tools for Natural Language Processing! What's your favorite?"
)
print(f"Original Text:\n{sample_text}")
```

## Sentence Tokenization with NLTK

```python
nltk_sentences = nltk.sent_tokenize(sample_text)
print("NLTK Sentence Tokenization:")
for i, sent in enumerate(nltk_sentences):
    print(f"{i+1}. {sent}")
```

## Word Tokenization with NLTK

```python
nltk_words = nltk.word_tokenize(sample_text)
print("NLTK Word Tokenization:")
print(nltk_words)
```

## Tokenization with spaCy

```python
# Load the English spaCy model
nlp = spacy.load('en_core_web_sm')

# Process the text with spaCy
doc = nlp(sample_text)

# SpaCy Sentence Tokenization
spacy_sentences = [sent.text for sent in doc.sents]
print("\nspaCy Sentence Tokenization:")
for i, sent in enumerate(spacy_sentences):
    print(f"{i+1}. {sent}")

# SpaCy Word Tokenization
spacy_words = [token.text for token in doc]
print("\nspaCy Word Tokenization:")
print(spacy_words)
```
