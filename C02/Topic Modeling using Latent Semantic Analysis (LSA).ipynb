```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# ---------------------------------------------------
# 1. Sample documents
# ---------------------------------------------------

documents = [
    "The team won the cricket match with excellent batting",
    "The cricket player scored a century in the match",
    "Football players trained hard for the championship",
    "The football team won the tournament",

    "The computer uses a powerful processor and memory",
    "Artificial intelligence and machine learning are growing",
    "The software uses machine learning algorithms",
    "Computer technology is developing rapidly"
]

# ---------------------------------------------------
# 2. Convert documents into TF-IDF matrix
# ---------------------------------------------------

vectorizer = TfidfVectorizer(stop_words="english")

X = vectorizer.fit_transform(documents)

print("TF-IDF Matrix Shape:", X.shape)

# ---------------------------------------------------
# 3. Apply LSA using Truncated SVD
# ---------------------------------------------------

num_topics = 2

lsa = TruncatedSVD(
    n_components=num_topics,
    random_state=42
)

X_lsa = lsa.fit_transform(X)

# ---------------------------------------------------
# 4. Display discovered topics
# ---------------------------------------------------

terms = vectorizer.get_feature_names_out()

print("\n========== DISCOVERED TOPICS ==========\n")

for topic_number, component in enumerate(lsa.components_):

    print("Topic", topic_number + 1)

    # Get top 5 words
    top_words = component.argsort()[-5:][::-1]

    for word_index in top_words:
        print(terms[word_index], end=" ")

    print("\n")

# ---------------------------------------------------
# 5. Display topic distribution for documents
# ---------------------------------------------------

print("========== DOCUMENT TOPICS ==========\n")

for i, topic_distribution in enumerate(X_lsa):

    topic = topic_distribution.argmax() + 1

    print("Document", i + 1)
    print(documents[i])
    print("Main Topic:", topic)
    print("Topic Scores:", topic_distribution)
    print("-" * 60)

# ---------------------------------------------------
# 6. Explained variance
# ---------------------------------------------------

print("\nExplained Variance Ratio:")
print(lsa.explained_variance_ratio_)

print("Total Explained Variance:",
      lsa.explained_variance_ratio_.sum())
```
