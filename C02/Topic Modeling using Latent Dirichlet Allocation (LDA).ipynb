```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# ---------------------------------------------------
# 1. Sample documents
# ---------------------------------------------------

documents = [
    "The team won the cricket match with excellent batting",
    "The cricket player scored a century in the match",
    "Football players trained hard for the championship",
    "The football team won the tournament",

    "The computer uses a powerful processor and memory",
    "Artificial intelligence and machine learning are growing",
    "The software uses machine learning algorithms",
    "Computer technology is developing rapidly"
]

# ---------------------------------------------------
# 2. Convert text into Bag-of-Words
# ---------------------------------------------------

vectorizer = CountVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(documents)

# ---------------------------------------------------
# 3. Create LDA model
# ---------------------------------------------------

num_topics = 2

lda_model = LatentDirichletAllocation(
    n_components=num_topics,
    random_state=42
)

lda_model.fit(X)

# ---------------------------------------------------
# 4. Display topics
# ---------------------------------------------------

words = vectorizer.get_feature_names_out()

print("========== DISCOVERED TOPICS ==========\n")

for topic_number, topic in enumerate(lda_model.components_):

    print("Topic", topic_number + 1)

    # Get the 5 most important words
    top_words = topic.argsort()[-5:][::-1]

    for word_index in top_words:
        print(words[word_index], end=" ")

    print("\n")

# ---------------------------------------------------
# 5. Assign topics to documents
# ---------------------------------------------------

document_topics = lda_model.transform(X)

print("========== DOCUMENT TOPICS ==========\n")

for i, topic_distribution in enumerate(document_topics):

    topic = topic_distribution.argmax() + 1

    print("Document", i + 1)
    print(documents[i])
    print("Main Topic:", topic)
    print("Topic Distribution:", topic_distribution)
    print("-" * 60)
```
