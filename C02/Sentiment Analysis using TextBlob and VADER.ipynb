```python
from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Create VADER analyzer
analyzer = SentimentIntensityAnalyzer()

# Sample texts
texts = [
    "I love this product. It is absolutely amazing!",
    "This movie is very good and enjoyable.",
    "The service was okay.",
    "I don't like this product.",
    "This movie was terrible and boring!"
]

print("========== SENTIMENT ANALYSIS ==========\n")

for text in texts:

    # -------------------------------
    # TextBlob
    # -------------------------------
    blob = TextBlob(text)
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0:
        tb_sentiment = "Positive"
    elif polarity < 0:
        tb_sentiment = "Negative"
    else:
        tb_sentiment = "Neutral"

    # -------------------------------
    # VADER
    # -------------------------------
    scores = analyzer.polarity_scores(text)
    compound = scores["compound"]

    if compound >= 0.05:
        vader_sentiment = "Positive"
    elif compound <= -0.05:
        vader_sentiment = "Negative"
    else:
        vader_sentiment = "Neutral"

    # -------------------------------
    # Display results
    # -------------------------------
    print("Text:", text)

    print("TextBlob Polarity:", polarity)
    print("TextBlob Subjectivity:", subjectivity)
    print("TextBlob Sentiment:", tb_sentiment)

    print("VADER Scores:", scores)
    print("VADER Compound:", compound)
    print("VADER Sentiment:", vader_sentiment)

    print("-" * 60)
```

```python
text = input("Enter a sentence: ")

# TextBlob
blob = TextBlob(text)
polarity = blob.sentiment.polarity

if polarity > 0:
    tb_result = "Positive"
elif polarity < 0:
    tb_result = "Negative"
else:
    tb_result = "Neutral"

# VADER
scores = analyzer.polarity_scores(text)
compound = scores["compound"]

if compound >= 0.05:
    vader_result = "Positive"
elif compound <= -0.05:
    vader_result = "Negative"
else:
    vader_result = "Neutral"

print("\nText:", text)
print("TextBlob:", tb_result)
print("VADER:", vader_result)
print("VADER Scores:", scores)
```
