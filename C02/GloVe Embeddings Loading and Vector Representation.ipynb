```python
import os
import urllib.request
import zipfile

# Download GloVe embeddings
url = "https://nlp.stanford.edu/data/glove.6B.zip"
zip_file = "glove.6B.zip"

urllib.request.urlretrieve(url, zip_file)

# Extract files
with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall("glove")

print("GloVe downloaded successfully!")
```

```python
import numpy as np

def load_glove(file_path):
    embeddings = {}

    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            values = line.split()
            word = values[0]
            vector = np.asarray(values[1:], dtype='float32')
            embeddings[word] = vector

    return embeddings


# Correct file path
glove_file = "glove/glove.6B.100d.txt"

# Load GloVe
glove_embeddings = load_glove(glove_file)

print("Number of words:", len(glove_embeddings))

# Get vector representation
word = "language"

if word in glove_embeddings:
    vector = glove_embeddings[word]

    print("\nWord:", word)
    print("Vector:")
    print(vector)

    print("\nVector Dimension:", len(vector))
else:
    print("Word not found.")
```
