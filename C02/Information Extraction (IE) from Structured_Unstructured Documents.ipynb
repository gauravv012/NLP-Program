```python
import re

# ---------------------------------------------------
# 1. Unstructured document
# ---------------------------------------------------

document = """
John Smith purchased a Dell Laptop from ABC Technologies
on 25 September 2026. The laptop costs Rs. 75,000.
For more information contact John at john.smith@gmail.com
or call +91-9876543210.

The company is located in New Delhi.
"""

# ---------------------------------------------------
# 2. Information Extraction using Regular Expressions
# ---------------------------------------------------

# Extract Email
email = re.findall(
    r'[\w\.-]+@[\w\.-]+\.\w+',
    document
)

# Extract Phone Number
phone = re.findall(
    r'(?:\+91[-\s]?)?[6-9]\d{9}',
    document
)

# Extract Date
date = re.findall(
    r'\b\d{1,2}\s+(?:January|February|March|April|May|June|'
    r'July|August|September|October|November|December)\s+\d{4}\b',
    document
)

# Extract Price
price = re.findall(
    r'(?:Rs\.?|INR)\s?[\d,]+',
    document
)

# Extract Person Name
name = re.findall(
    r'\b[A-Z][a-z]+\s[A-Z][a-z]+\b',
    document
)

# Extract Organization
organization = re.findall(
    r'\b(?:ABC Technologies|Dell)\b',
    document
)

# ---------------------------------------------------
# 3. Display extracted information
# ---------------------------------------------------

print("========== INFORMATION EXTRACTION ==========\n")

print("Names:")
for item in name:
    print("-", item)

print("\nOrganizations:")
for item in organization:
    print("-", item)

print("\nEmail:")
for item in email:
    print("-", item)

print("\nPhone:")
for item in phone:
    print("-", item)

print("\nDate:")
for item in date:
    print("-", item)

print("\nPrice:")
for item in price:
    print("-", item)
```
