```python
import nltk
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

# Input texts
text1 = "The cat sits on the mat"
text2 = "A kitten is resting on a rug"

# Tokenization
words1 = word_tokenize(text1.lower())
words2 = word_tokenize(text2.lower())

# Training corpus
corpus = [
    words1,
    words2,
    ["cat", "kitten", "animal"],
    ["sits", "resting", "sleeping"],
    ["mat", "rug", "floor"]
]

# Train Word2Vec
model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    seed=42
)

# Calculate WMD
distance = model.wv.wmdistance(words1, words2)

# Similarity score
similarity = 1 / (1 + distance)

print("Text 1:", text1)
print("Text 2:", text2)
print("WMD:", distance)
print("Similarity Score:", similarity)
```
