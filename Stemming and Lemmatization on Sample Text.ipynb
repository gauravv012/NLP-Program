## Stemming and Lemmatization

First, we'll import the necessary NLTK libraries and download the required data. If you run into issues, make sure you have an internet connection.

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download necessary NLTK data (run once)
# 'punkt' for tokenization
# 'wordnet' for lemmatization
# 'omw-1.4' for WordNetLemmatizer
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet')
try:
    nltk.data.find('corpora/omw-1.4')
except LookupError:
    nltk.download('omw-1.4')

print("NLTK resources downloaded successfully!")
```

Now, let's define a sample text, tokenize it, and then apply both stemming and lemmatization.

```python
# Download punkt_tab if not already present (required for word_tokenize)
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

# Sample text
sample_text = "Running, runs, runner, ran are all forms of the verb to run. Similarly, dogs, cats, and birds are plurals."

# Tokenize the text
words = word_tokenize(sample_text)
print(f"Original words: {words}\n")

# Initialize stemmer and lemmatizer
porter_stemmer = PorterStemmer()
wordnet_lemmatizer = WordNetLemmatizer()

# Perform stemming
stemmed_words = [porter_stemmer.stem(word) for word in words]
print(f"Stemmed words: {stemmed_words}\n")

# Perform lemmatization
# For better results, lemmatization often requires part-of-speech (POS) tagging.
# Here, we'll use a default POS for simplicity (noun).
# If you want more accurate lemmatization, consider using POS tags.
lemmatized_words = [wordnet_lemmatizer.lemmatize(word) for word in words]
print(f"Lemmatized words (default POS=noun): {lemmatized_words}\n")

# Example with specific POS tag for 'running' (verb)
lemmatized_verb = wordnet_lemmatizer.lemmatize("running", pos="v")
print(f"Lemmatized 'running' (as verb): {lemmatized_verb}")

lemmatized_plural = wordnet_lemmatizer.lemmatize("dogs", pos="n")
print(f"Lemmatized 'dogs' (as noun): {lemmatized_plural}")
```
